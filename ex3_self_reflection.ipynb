Exercise 3: Self-Reflection Prompt (Critique + Improve Summary)
Tools used: ChatGPT web + Colab

BEFORE SUMMARY:
“We fixed the issue and everything is good now. It was probably caused by a system glitch. You shouldn’t experience this problem again and there’s nothing else you need to do. Let us know if something happens.”

System:
You are a strict editor. Critique first, then rewrite.

User:
Here is my summary:
"We fixed the issue and everything is good now. It was probably caused by a system glitch. You shouldn’t experience this problem again and there’s nothing else you need to do. Let us know if something happens."

Requirements:

Audience: non-technical customer

Max 90 words

Must include:

1-sentence main point

2 bullet takeaways

1 next step

Avoid jargon, blaming, and absolute guarantees

Do not add new facts

Do:

CRITIQUE: 4–6 concrete issues

REWRITE: improved summary meeting all requirements

CRITIQUE

The main point is vague (“everything is good now”) and not clearly centered in one strong sentence.

“System glitch” is unclear and mildly technical without explanation.

“You shouldn’t experience this problem again” sounds like a guarantee.

“There’s nothing else you need to do” is absolute and risky.

The structure is a paragraph, not organized with clear takeaways or a next step.

REWRITE

The issue has been resolved, and your account is working normally again.

The problem appears to have been temporary.

No action is required from you at this time.

Next step: Please continue using the service as usual and let us know if you notice anything unexpected.
